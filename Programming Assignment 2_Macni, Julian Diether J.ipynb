## A. REPRODUCIBLE NORMALIZATION PROBLEM

```python
#gets a random set of numbers and makes an array
np.random.seed(2112)
X = np.random.randint(10, 101, size=(5, 5))
X
```

```python
#gets the normalized array
X_normalized = (X - np.mean(X)) / np.std(X)
X_normalized
```

```python
#gets the mean of the normalized array
mean = np.mean(X_normalized)
mean
```

```python
#gets the standard deviation of the normalized array
standard_deviation = np.std(X_normalized)
standard_deviation
```

```python
#saves the normalized array
np.save("X_normalized.npy", X_normalized)
```

## B. CUBES DIVISIBLE BY 4 PROBLEM

```python
#makes a 10 x 10 array of the cubes of the first 100 positive integers 
C = np.arange(1, 101, 1).reshape(10, 10) ** 3
C
```

```python
#displays the shape of the array
C.shape
```

```python
#gets the array of integers divisible by 4
div_by_4 = C[C % 4 == 0]
div_by_4
```

```python
#displays the number of elements in the array
div_by_4.size
```

```python
#saves the array
np.save("div_by_4.npy", div_by_4)
```

## C. ABOVE-MEANS SQUARE PROBLEM

```python
#makes a 6 x 6 array of the squares of the first 36 positive integers
S = np.arange(1, 37, 1).reshape(6, 6) ** 2
S
```

```python
#gets the mean of the array
S_mean = np.mean(S)
S_mean
```

```python
#makes an array of the elements above the mean
above_mean = S[S > S_mean]
above_mean
```

```python
#displays the number of elements in the array
above_mean.size
```

```python
#saves the array
np.save("above_mean.npy", above_mean)
```
